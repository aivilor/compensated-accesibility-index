```python

def get_2sfca (ciudad, t):

    dfs = None
    for loc in [1, 2, 3, 4]:
        col_name = "{}_loc{}_t{}".format(ciudad, loc, t)
        path = "/Users/aidavillalba/Desktop/Artículos/2026/02_Mediterranean cities/resultados/{0}/2sfca_loc{1}_t{2}_{0}.geojson".format(ciudad, loc, t)
        df_resultados = gpd.read_file(path)
        df_resultados = df_resultados.rename(columns={"Aj_Sum": col_name})
        if dfs is None:
            # first iteration: keep id, geometry, and the value column
            dfs = df_resultados[["nationalCa", "geometry", col_name]].copy()
        else:
            # subsequent iterations: merge in only id + value column (no duplicate geometry)
            dfs = dfs.merge(df_resultados[["nationalCa", col_name]],how="left",on="nationalCa")
        dfs.to_file("/Users/aidavillalba/Desktop/Artículos/2026/02_Mediterranean cities/resultados/{}/{}_2sfca.geojson".format(ciudad, ciudad))
```

```python
get_2sfca ("Barcelona", 15)
get_2sfca ("Valencia", 15)
get_2sfca ("Murcia", 15)
```
